# 世界通史时空图谱 · 时间轴连续演进「编年快报」语音工坊（Colab）

为底部播放键（`▷` 连续演进）与「编年快报」弹窗生成全部 **202 条精炼快报旁白**（8 大时代 + 24 个细分阶段 + 170 项重大史事），每条约 4~6 秒、32kbps 轻量化 MP3（整包仅约 **5 MB**）。

- 默认并发 `CONCURRENCY = 6`，全量 **202 条** 约 **1~2 分钟** 即完成，且支持标准 CPU 运行时（不消耗 GPU 计算点）。
- 回到本地运行：
```bash
npm run media:import -- ~/Downloads/atlas-ticker-audio-pack.zip
```

````python
#@title 1. 设置工作目录（无需 GPU，可选挂载 Drive）
import os, shutil

USE_GOOGLE_DRIVE = False  #@param {type:"boolean"}
DRIVE_SUBDIR = "world-history-atlas/ticker-audio"  #@param {type:"string"}

WORK_DIR = "/content/atlas-ticker-audio"
if USE_GOOGLE_DRIVE:
    try:
        from google.colab import drive
        drive.mount("/content/drive")
        WORK_DIR = os.path.join("/content/drive/MyDrive", DRIVE_SUBDIR)
    except Exception as exc:
        print(f"Drive 挂载跳过，使用本地目录: {exc}")

os.makedirs(WORK_DIR, exist_ok=True)
print("📁 工作目录:", WORK_DIR, "| ffmpeg:", shutil.which("ffmpeg") or "待安装")
````

````python
#@title 2. 安装依赖与参数配置 (edge-tts + ffmpeg)
VOICE_PRESET = "zh-CN-YunjianNeural"  #@param ["zh-CN-YunjianNeural", "zh-CN-YunxiNeural", "zh-CN-XiaoxiaoNeural"]
SPEECH_RATE = "+8%"  #@param {type:"string"}
MP3_BITRATE = "32k"  #@param ["24k", "32k", "48k"]
CONCURRENCY = 6  #@param {type:"integer"}

!apt-get -qq update && apt-get -qq install -y ffmpeg
!pip -q install --upgrade edge-tts mutagen nest_asyncio
print(f"✅ 依赖就绪 | 音色: {VOICE_PRESET} | 语速: {SPEECH_RATE} | 并发: {CONCURRENCY}")
````

````python
#@title 3. 拉取时间轴快报脚本清单 (ticker-audio-scripts.json)
import json, urllib.request

RAW_SCRIPTS_URL = "https://raw.githubusercontent.com/quanliulq-pixel/world-history-atlas/main/colab/ticker-audio-scripts.json"  #@param {type:"string"}
LOCAL_SCRIPTS_PATH = os.path.join(WORK_DIR, "ticker-audio-scripts.json")

urllib.request.urlretrieve(RAW_SCRIPTS_URL, LOCAL_SCRIPTS_PATH)
with open(LOCAL_SCRIPTS_PATH, "r", encoding="utf-8") as f:
    ticker_catalog = json.load(f)

print("📊 快报清单统计:", ticker_catalog.get("counts", {}))
````

````python
#@title 4. 并发批量合成 202 条编年快报音频并生成 manifest.json
import asyncio, datetime, subprocess, time
import nest_asyncio
import edge_tts
from mutagen.mp3 import MP3
from IPython.display import Audio, display

nest_asyncio.apply()

MANIFEST_PATH = os.path.join(WORK_DIR, "manifest.json")
manifest = {"version": 1, "engine": f"edge-tts:{VOICE_PRESET}", "items": {}}
items = ticker_catalog["items"]

def get_duration(mp3_path: str) -> float:
    try:
        return round(float(MP3(mp3_path).info.length), 2)
    except Exception:
        return 0.0

async def synth_one(sem: asyncio.Semaphore, idx: int, item: dict, stats: dict):
    rel_file = item["file"]
    abs_file = os.path.join(WORK_DIR, rel_file)
    os.makedirs(os.path.dirname(abs_file), exist_ok=True)

    async with sem:
        if not os.path.exists(abs_file):
            tmp_mp3 = abs_file + ".raw.mp3"
            for attempt in range(3):
                try:
                    comm = edge_tts.Communicate(item["script"], VOICE_PRESET, rate=SPEECH_RATE)
                    await comm.save(tmp_mp3)
                    subprocess.run(
                        ["ffmpeg", "-y", "-loglevel", "error", "-i", tmp_mp3, "-ac", "1", "-ar", "24000", "-b:a", MP3_BITRATE, abs_file],
                        check=True,
                    )
                    if os.path.exists(tmp_mp3):
                        os.remove(tmp_mp3)
                    break
                except Exception as exc:
                    if attempt == 2:
                        raise exc
                    await asyncio.sleep(1.2 * (attempt + 1))

    manifest["items"][item["key"]] = {
        "kind": "ticker",
        "id": item["id"],
        "script": item["script"],
        "voice": VOICE_PRESET,
        "file": rel_file,
        "durationSec": get_duration(abs_file),
    }
    stats["done"] += 1
    if stats["done"] % 25 == 0 or stats["done"] == len(items):
        elapsed = time.time() - stats["start"]
        print(f"[{stats['done']}/{len(items)}] ✅ {item['title']} | 已耗时 {elapsed:.1f}s")

async def run_all():
    sem = asyncio.Semaphore(max(1, CONCURRENCY))
    stats = {"done": 0, "start": time.time()}
    await asyncio.gather(*(synth_one(sem, idx, item, stats) for idx, item in enumerate(items)))
    manifest["generatedAt"] = datetime.datetime.now(datetime.timezone.utc).isoformat().replace("+00:00", "Z")
    with open(MANIFEST_PATH, "w", encoding="utf-8") as f:
        json.dump(manifest, f, ensure_ascii=False, indent=2)
    print(f"\n🎉 全部 {len(manifest['items'])} 条快报音频合成完成！")
    first_abs = os.path.join(WORK_DIR, items[0]["file"])
    if os.path.exists(first_abs):
        display(Audio(first_abs))

asyncio.get_event_loop().run_until_complete(run_all())
````

````python
#@title 5. 打包导出 atlas-ticker-audio-pack.zip 并自动下载
import zipfile

ZIP_PATH = "/content/atlas-ticker-audio-pack.zip"
file_count = 0

with zipfile.ZipFile(ZIP_PATH, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    zf.write(MANIFEST_PATH, "manifest.json")
    for entry in manifest["items"].values():
        rel = entry["file"]
        abs_path = os.path.join(WORK_DIR, rel)
        if os.path.exists(abs_path):
            zf.write(abs_path, rel)
            file_count += 1

size_mb = os.path.getsize(ZIP_PATH) / (1024 * 1024)
print(f"📦 已打包 {file_count} 段快报音频 -> {ZIP_PATH} ({size_mb:.2f} MB)")

try:
    from google.colab import files
    files.download(ZIP_PATH)
except Exception as exc:
    print(f"请手动下载 {ZIP_PATH}: {exc}")
````
